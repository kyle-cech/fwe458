```python
import random
import copy
```

```python
lis = random.sample(range(1, 101), 50) # initialize unsorted list lis
def insert_sort(lis):
  sorted_lis = [] # initialize empty list sorted_lis
  l = copy.copy(lis) # create a copy of lis, l, so that it won't be affected by popping elements
  while len(l) > 0:
    idx = l.index(min(l)) # find index of lowest value in original list l
    sorted_lis.append(l[idx]) # add lowest value to sorted_lis
    l.pop(idx) # remove lowest number from l
  return sorted_lis # return sorted list
```

```python
def bubble_sort(lis):
  l = copy.copy(lis) # create a copy of lis, l, so that it won't be affected by sorting elements
  for i in range(len(l)):  # loop for as many times as there are items in l so every number is sorted
    for x in range(len(l) - i - 1): # for every item in l, loop over every item after it (excluding the last number) to put it in the right spot
      if l[x] > l[x + 1]:
        l[x + 1], l[x] = l[x], l[x + 1] # if a number is greater than the number to the right of it, switch their positions in l
  return l # return l which is now sorted
```

```python
# Testing
insert = insert_sort(lis)
bubble = bubble_sort(lis)
print(lis)
print(insert)
print(bubble)
print(insert == bubble) # check if both methods get same result
```
